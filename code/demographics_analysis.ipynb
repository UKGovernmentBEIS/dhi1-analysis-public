# Not runnable against the public dataset

This notebook computes the sample-composition table from the 12 participant
demographics in `Ztable_indiv_combined.csv`. Those columns are **removed from the
public release** (see the Data section of the README), so no cell below will run as published.

Its output, `data/demographics_aggregated.csv`, is committed and is safe to use — it
contains marginal frequencies only, with no cross-tabulations.

The notebook is retained to document how that table was derived. Re-running it
requires controlled access to the underlying data.

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import re
```

```python
data = pd.read_csv('../data/Ztable_indiv_combined.csv')

print(len(data))
data.head(10)
```

```python
data.columns
```

```python
categorical_variables = ['age', 'gender', 'ethnicity', 'region', 'income', 'religion', 'education', 'vote', 'brexit', 'disabled', 'mental_health', 'chatbot_use']

# Initialize an empty dictionary to store the results
results = {}

for var in categorical_variables:
    frequency = data[var].value_counts()
    percentage = data[var].value_counts(normalize=True) * 100
    results[var] = pd.DataFrame({'Frequency': frequency, 'Percentage': percentage})

# Combine results into a single dataframe for convenience
combined_results = pd.concat(results, axis=1)
combined_results

combined_results.to_csv('demographics_aggregated.csv')
```

```python
# Helper function to sort age groups numerically
def sort_age_groups(age_group):
    match = re.match(r'(\d+)-(\d+)', age_group)
    if match:
        return int(match.group(1))
    return age_group


# Initialize an empty dictionary to store the results
# Initialize an empty dictionary to store the results
results = {}

for var in categorical_variables:
    percentage = data[var].value_counts(normalize=True) * 100
    results[var] = pd.DataFrame({'Percentage': percentage})

# Bullet list formatting
output = []

for var in categorical_variables:
    output.append(f"{var.capitalize()}:")
    if var == 'age':
        sorted_index = sorted(results[var].index, key=sort_age_groups)
    else:
        sorted_index = results[var].sort_values('Percentage', ascending=False).index
    
    for index in sorted_index:
        output.append(f"  - {index}: {results[var].loc[index, 'Percentage']:.2f}\%")

# Print the bullet list
for line in output:
    print(line)
```
